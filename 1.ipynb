```python
from collections import Counter
import matplotlib.pyplot as plt

def participation_frequency_boxplot(data, methods=None, num_clients=30, ax=None):
    """
    Create a box plot of per-client participation frequencies.

    For each method:
      1. Counts each client's participation over 10 rounds within each seed.
      2. Averages those counts across seeds.
      3. Plots the distribution across the 30 clients.

    The y-axis therefore ranges from 0 to the number of rounds per seed
    (normally 10), rather than 0 to 30.
    """
    if methods is None:
        methods = list(data.keys())

    frequencies_by_method = {}

    for method in methods:
        per_seed_frequencies = []

        # Each item in data[method] is one seed/run
        for run in data[method]:
            counts = Counter()

            # Each record represents one round/iteration
            for record in run.values():
                for client_id in record.get("participating_clients", []):
                    counts[int(client_id)] += 1

            # Frequency vector for all 30 clients in this seed
            seed_frequency = [counts[i] for i in range(num_clients)]
            per_seed_frequencies.append(seed_frequency)

        # Average frequency for each client over all seeds
        frequencies_by_method[method] = np.mean(
            per_seed_frequencies,
            axis=0
        )

    if ax is None:
        fig, ax = plt.subplots(figsize=(9, 5))

    ax.boxplot(
        [frequencies_by_method[method] for method in methods],
        labels=["NSGA-II", "Full Selection", "Random 75%", "Random 50%", "Random 25%"],
        showfliers=True
    )

    ax.set_xlabel("Method")
    ax.set_ylabel("Average participation rounds per client")
    ax.set_title("Distribution of client participation frequency")
    ax.set_ylim(0, 10)
    ax.grid(axis="y", linestyle="--", alpha=0.3)

    return ax, frequencies_by_method

participation_frequency_boxplot(all_methods_no_warmup)
```

```python
import numpy as np
import matplotlib.pyplot as plt


def plot_selected_vs_participating_utility(
    data,
    methods=None,
    method_labels=None,
    ax_array=None,
    show_variability=True,
    variability="std",
    figsize=(13, 7),
):
    """
    Plot average utility of selected clients versus actual participating clients
    over communication rounds, averaged across seeds/runs.

    Parameters
    ----------
    data : dict
        JSON-loaded experiment data. Expected structure:

        data[method][run_index][round_number] = {
            "average_utility_selected_clients_before_fit": float,
            "average_utility_participating_clients_before_fit": float,
            ...
        }

    methods : list of str, optional
        Methods to plot. By default, uses:
        ["nsga2", "all", "75p", "50p", "25p"]
        when available.

    method_labels : dict, optional
        Display names for methods, for example:
        {"nsga2": "NSGA-II", "75p": "Random 75%"}

    ax_array : array-like of matplotlib Axes, optional
        Existing axes to draw on. If omitted, a new subplot grid is created.

    show_variability : bool
        If True, draw a shaded uncertainty band across runs/seeds.

    variability : {"std", "sem"}
        Type of shaded band:
        - "std": ± one standard deviation across seeds
        - "sem": ± standard error of the mean across seeds

    figsize : tuple
        Size of the complete figure when axes are created internally.

    Returns
    -------
    fig : matplotlib.figure.Figure
    axes : numpy.ndarray
        Array of axes containing the method panels.
    results : dict
        Per-method averaged curves and raw per-seed values.
    """

    default_methods = ["nsga2", "all", "75p", "50p", "25p"]

    if methods is None:
        methods = [m for m in default_methods if m in data]

        # Include any extra methods that may exist in the JSON.
        methods += [m for m in data if m not in methods]

    if method_labels is None:
        method_labels = {
            "nsga2": "NSGA-II",
            "all": "Full Selection",
            "75p": "Random 75%",
            "50p": "Random 50%",
            "25p": "Random 25%",
        }

    # Build a 2-by-3 layout for five methods.
    if ax_array is None:
        fig, axes = plt.subplots(
            nrows=2,
            ncols=3,
            figsize=figsize,
            sharex=True,
            sharey=True,
            constrained_layout=True,
        )
        axes = np.asarray(axes).ravel()
    else:
        axes = np.asarray(ax_array).ravel()
        fig = axes[0].figure

    if len(axes) < len(methods):
        raise ValueError(
            f"Need at least {len(methods)} axes, but only {len(axes)} were provided."
        )

    results = {}

    for ax, method in zip(axes, methods):
        runs = data[method]

        # Obtain and sort round IDs numerically, e.g. "1", "2", ..., "10".
        round_keys = sorted(runs[0].keys(), key=lambda x: int(x))
        rounds = np.array([int(round_key) for round_key in round_keys])

        # Shape: (number_of_seeds, number_of_rounds)
        selected_utilities = np.array([
            [
                run[round_key]["average_utility_selected_clients_before_fit"]
                for round_key in round_keys
            ]
            for run in runs
        ], dtype=float)

        participating_utilities = np.array([
            [
                run[round_key]["average_utility_participating_clients_before_fit"]
                for round_key in round_keys
            ]
            for run in runs
        ], dtype=float)

        selected_mean = np.nanmean(selected_utilities, axis=0)
        participating_mean = np.nanmean(participating_utilities, axis=0)

        if variability == "std":
            selected_spread = np.nanstd(selected_utilities, axis=0)
            participating_spread = np.nanstd(participating_utilities, axis=0)

        elif variability == "sem":
            n_selected = np.sum(~np.isnan(selected_utilities), axis=0)
            n_participating = np.sum(~np.isnan(participating_utilities), axis=0)

            selected_spread = (
                np.nanstd(selected_utilities, axis=0, ddof=1)
                / np.sqrt(n_selected)
            )
            participating_spread = (
                np.nanstd(participating_utilities, axis=0, ddof=1)
                / np.sqrt(n_participating)
            )

        else:
            raise ValueError("variability must be either 'std' or 'sem'.")

        # Selected clients: solid blue line.
        ax.plot(
            rounds,
            selected_mean,
            color="#1f77b4",
            linewidth=2.2,
            linestyle="-",
            marker="o",
            markersize=4,
            label="Selected clients",
        )

        # Actual participants: dashed orange line.
        ax.plot(
            rounds,
            participating_mean,
            color="#ff7f0e",
            linewidth=2.2,
            linestyle="--",
            marker="s",
            markersize=4,
            label="Participating clients",
        )

        if show_variability:
            ax.fill_between(
                rounds,
                selected_mean - selected_spread,
                selected_mean + selected_spread,
                color="#1f77b4",
                alpha=0.15,
                linewidth=0,
            )

            ax.fill_between(
                rounds,
                participating_mean - participating_spread,
                participating_mean + participating_spread,
                color="#ff7f0e",
                alpha=0.15,
                linewidth=0,
            )

        ax.set_title(method_labels.get(method, method), fontweight="bold")
        ax.set_xticks(rounds)
        ax.grid(axis="y", linestyle="--", alpha=0.3)

        results[method] = {
            "rounds": rounds,
            "selected_mean": selected_mean,
            "participating_mean": participating_mean,
            "selected_spread": selected_spread,
            "participating_spread": participating_spread,
            "selected_by_seed": selected_utilities,
            "participating_by_seed": participating_utilities,
        }

    # Hide unused axes, e.g. sixth panel in the 2×3 grid.
    for ax in axes[len(methods):]:
        ax.set_visible(False)

    # Shared labels.
    for ax in axes[:len(methods)]:
        ax.set_xlabel("Communication round")
        ax.set_ylabel("Average utility")

    # One common legend for the full figure.
    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(
        handles,
        labels,
        loc="upper center",
        ncol=2,
        frameon=False,
        bbox_to_anchor=(0.5, 1.03),
    )

    return fig, axes, results



fig, axes, utility_results = plot_selected_vs_participating_utility(
    all_methods_no_warmup,
    variability="std",      # Or use "sem"
    show_variability=True,
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


def plot_global_test_accuracy_with_warmup(
    data,
    methods=None,
    method_labels=None,
    show_percent=False,
    figsize=(9, 5),
    ax=None,
):
    """
    Plot global test accuracy across warm-up + 10 optimized rounds.

    For each method:
    - x-axis: Warm-up (Round 0) + rounds 1..10
    - y-axis: global test accuracy
    - line: mean across 3 runs
    - shaded band: ± standard deviation across 3 runs

    Parameters
    ----------
    data : dict
        JSON-loaded experiment data from all_methods.json.

    methods : list[str], optional
        Methods to plot. Defaults to the standard five methods in file order.

    method_labels : dict[str, str], optional
        Display labels for methods.

    show_percent : bool
        If True, multiply accuracy by 100 and label as percent.

    figsize : tuple
        Figure size if ax is not provided.

    ax : matplotlib.axes.Axes, optional
        Existing axis to draw on.

    Returns
    -------
    fig, ax, results
    """
    default_methods = ["nsga2", "all", "75p", "50p", "25p"]

    if methods is None:
        methods = [m for m in default_methods if m in data]

    if method_labels is None:
        method_labels = {
            "nsga2": "NSGA-II",
            "all": "All clients",
            "75p": "Random 75%",
            "50p": "Random 50%",
            "25p": "Random 25%",
        }

    if ax is None:
        fig, ax = plt.subplots(figsize=figsize)
    else:
        fig = ax.figure

    results = {}

    colors = {
        "nsga2": "#1f77b4",
        "all": "#2ca02c",
        "75p": "#ff7f0e",
        "50p": "#d62728",
        "25p": "#9467bd",
    }

    for method in methods:
        runs = data[method]

        # Enforce the intended order: warm-up first, then rounds 1..10
        round_keys = ["warm_up"] + [str(i) for i in range(1, 11)]
        rounds = np.arange(0, 11)

        accuracy_by_run = np.array([
            [run[rk]["global_test_accuracy"] for rk in round_keys]
            for run in runs
        ], dtype=float)

        if show_percent:
            accuracy_by_run = accuracy_by_run * 100.0

        mean_accuracy = np.mean(accuracy_by_run, axis=0)
        std_accuracy = np.std(accuracy_by_run, axis=0)

        color = colors.get(method, None)
        label = method_labels.get(method, method)

        ax.plot(
            rounds,
            mean_accuracy,
            marker="o",
            linewidth=2,
            markersize=4,
            color=color,
            label=label,
        )

        ax.fill_between(
            rounds,
            mean_accuracy - std_accuracy,
            mean_accuracy + std_accuracy,
            color=color,
            alpha=0.15,
            linewidth=0,
        )

        results[method] = {
            "rounds": rounds,
            "mean": mean_accuracy,
            "std": std_accuracy,
            "values_by_run": accuracy_by_run,
        }

    ax.set_xlabel("Communication round")
    ax.set_xticks(np.arange(0, 11))
    ax.set_xticklabels(["Warm-up"] + [str(i) for i in range(1, 11)])

    if show_percent:
        ax.set_ylabel("Global test accuracy (%)")
    else:
        ax.set_ylabel("Global test accuracy")

    ax.set_title("Global test accuracy over communication rounds")
    ax.grid(axis="y", linestyle="--", alpha=0.3)
    ax.legend(frameon=False)

    if show_percent:
        ax.set_ylim(0, 100)
    else:
        ax.set_ylim(0, 1)

    fig.tight_layout()
    return fig, ax, results


fig, ax, results = plot_global_test_accuracy_with_warmup(
    all_methods,
    show_percent=False
)
```

```python
import numpy as np
import matplotlib.pyplot as plt


def plot_fairness_over_rounds(
    data,
    methods=None,
    method_labels=None,
    show_warmup=True,
    figsize=(9, 5),
    ax=None,
):
    """
    Plot fairness over communication rounds.

    For each method:
    - x-axis: Warm-up (Round 0) + rounds 1..10, or rounds 1..10 only
    - y-axis: fairness
    - line: mean across 3 runs
    - shaded band: ± standard deviation across 3 runs

    Parameters
    ----------
    data : dict
        JSON-loaded contents of all_methods.json.

    methods : list[str] or None
        Method keys to plot. Uses all standard methods by default.

    method_labels : dict[str, str] or None
        Friendly labels for plot legend.

    show_warmup : bool
        Include the `"warm_up"` record as round 0.

    figsize : tuple
        Figure size when `ax` is not supplied.

    ax : matplotlib.axes.Axes or None
        Optional existing axis to plot on.

    Returns
    -------
    fig : matplotlib.figure.Figure
    ax : matplotlib.axes.Axes
    results : dict
        Mean, standard deviation, and individual-run fairness values.
    """
    default_methods = ["nsga2", "all", "75p", "50p", "25p"]

    if methods is None:
        methods = [method for method in default_methods if method in data]

    if method_labels is None:
        method_labels = {
            "nsga2": "NSGA-II",
            "all": "All clients",
            "75p": "Random 75%",
            "50p": "Random 50%",
            "25p": "Random 25%",
        }

    colors = {
        "nsga2": "#1f77b4",
        "all": "#2ca02c",
        "75p": "#ff7f0e",
        "50p": "#d62728",
        "25p": "#9467bd",
    }

    if ax is None:
        fig, ax = plt.subplots(figsize=figsize)
    else:
        fig = ax.figure

    if show_warmup:
        round_keys = ["warm_up"] + [str(i) for i in range(1, 11)]
        rounds = np.arange(0, 11)
        x_tick_labels = ["Warm-up"] + [str(i) for i in range(1, 11)]
    else:
        round_keys = [str(i) for i in range(1, 11)]
        rounds = np.arange(1, 11)
        x_tick_labels = [str(i) for i in range(1, 11)]

    results = {}

    for method in methods:
        runs = data[method]

        # Rows = seeds/runs; columns = warm-up + communication rounds.
        fairness_by_run = np.array(
            [
                [run[round_key]["fairness"] for round_key in round_keys]
                for run in runs
            ],
            dtype=float,
        )

        mean_fairness = np.mean(fairness_by_run, axis=0)
        std_fairness = np.std(fairness_by_run, axis=0)

        color = colors.get(method)
        label = method_labels.get(method, method)

        ax.plot(
            rounds,
            mean_fairness,
            color=color,
            label=label,
            linewidth=2,
            marker="o",
            markersize=4,
        )

        ax.fill_between(
            rounds,
            mean_fairness - std_fairness,
            mean_fairness + std_fairness,
            color=color,
            alpha=0.15,
            linewidth=0,
        )

        results[method] = {
            "rounds": rounds,
            "mean": mean_fairness,
            "std": std_fairness,
            "values_by_run": fairness_by_run,
        }

    ax.set_xlabel("Communication round")
    ax.set_ylabel("Fairness")
    ax.set_title("Fairness over communication rounds")

    ax.set_xticks(rounds)
    ax.set_xticklabels(x_tick_labels)

    ax.grid(axis="y", linestyle="--", alpha=0.3)
    ax.legend(frameon=False)

    fig.tight_layout()

    return fig, ax, results

fig, ax, results = plot_fairness_over_rounds(
    all_methods,
    show_warmup=True
)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


def plot_pareto_front_pairwise(
    csv_path,
    round_number=5,
    generation=None,
    transform_to_natural_metrics=True,
    figsize=(14, 4.2),
):
    """
    Plot pairwise trade-offs between the 3 NSGA-II objectives for a representative round.

    Objectives (if transform_to_natural_metrics=True):
      - x/y 1: Average Client Utility (1 - objective_1) -> Higher is better
      - x/y 2: Selected Clients' Mean Accuracy (objective_2) -> Lower prioritizes struggling clients
      - x/y 3: Predicted Global Accuracy (1 - objective_3) -> Higher is better
    """
    df = pd.read_csv(csv_path)

    # Filter for the chosen communication round
    round_df = df[df["round"] == round_number].copy()
    if round_df.empty:
        raise ValueError(
            f"Round {round_number} not found. Available: {sorted(df['round'].unique())}"
        )

    # Pick the last generation in that round (the final Pareto front)
    if generation is None:
        generation = int(round_df["generation"].max())

    pop_df = round_df[round_df["generation"] == generation].copy()

    # Compute plotted values
    if transform_to_natural_metrics:
        # Convert minimization targets back to natural [0, 1] metrics
        pop_df["metric_utility"] = 1.0 - pop_df["objective_1"]
        pop_df["metric_fairness"] = pop_df["objective_2"]  # Lower = prioritizes worse clients
        pop_df["metric_accuracy"] = 1.0 - pop_df["objective_3"]

        x_cols = ["metric_utility", "metric_utility", "metric_fairness"]
        y_cols = ["metric_fairness", "metric_accuracy", "metric_accuracy"]

        labels = {
            "metric_utility": "Average Client Utility\n(Higher is better $\\uparrow$)",
            "metric_fairness": "Selected Clients' Mean Local Acc.\n(Lower $\\rightarrow$ prioritizes lagging clients)",
            "metric_accuracy": "Predicted Global Accuracy\n(Higher is better $\\uparrow$)",
        }
    else:
        # Raw internal minimization objectives (F1, F2, F3)
        x_cols = ["objective_1", "objective_1", "objective_2"]
        y_cols = ["objective_2", "objective_3", "objective_3"]

        labels = {
            "objective_1": "Objective 1: $1 - \\text{Utility}$ (Minimization)",
            "objective_2": "Objective 2: Prioritization (Minimization)",
            "objective_3": "Objective 3: $1 - \\text{Accuracy}$ (Minimization)",
        }

    fig, axes = plt.subplots(1, 3, figsize=figsize, constrained_layout=True)

    pairs = list(zip(x_cols, y_cols))

    for ax, (xc, yc) in zip(axes, pairs):
        # Sort values to draw a neat front outline if desired
        sub_df = pop_df.sort_values(by=xc)

        # Plot Pareto optimal points
        ax.scatter(
            sub_df[xc],
            sub_df[yc],
            color="#1f77b4",
            edgecolor="black",
            linewidth=0.8,
            s=60,
            zorder=3,
            label="Pareto solution",
        )

        ax.set_xlabel(labels[xc], fontsize=10)
        ax.set_ylabel(labels[yc], fontsize=10)
        ax.grid(True, linestyle="--", alpha=0.35)

    fig.suptitle(
        f"NSGA-II Non-Dominated Solutions (Trade-Offs) — Communication Round {round_number}",
        fontsize=12,
        fontweight="bold",
    )

    return fig, axes, pop_df




fig, axes, front_df = plot_pareto_front_pairwise(
    csv_path="version 7 - alpha0.1 runseed3 dataseed3 fashion mnist\\results\\nsga2\\nsga2_run_pareto.csv",
    round_number=5,
    transform_to_natural_metrics=True    # objective_directions=("min", "min", "min"),
    # objective_labels=(
    #     "Objective 1 - Utility (minimize)",
    #     "Objective 2 - Fairness (minimize)",
    #     "Objective 3 - Global Accuracy (minimize)",
    # ),
)

plt.show()
```
